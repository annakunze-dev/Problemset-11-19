## Challenge 5 — Separate a coefficient table from a predictive model

# 5. Newman–Raju governing equation for \(M_2\)

The selected coefficient is the Newman–Raju quadratic coefficient \(M_2\).

For the branch \(0<a/c\leq1\), the coefficient expressions used here are:

\[
M_1=1.13-0.09\left(\frac{a}{c}\right)
\]

\[
\boxed{M_2=-0.54+\frac{0.89}{0.2+(a/c)}}
\]

\[
M_3=0.5-\frac{1}{0.65+(a/c)}
+14\left(1-\frac{a}{c}\right)^{24}
\]

The polynomial geometry expression is:

\[
F=\left[
M_1+M_2\left(\frac{a}{t}\right)^2
+M_3\left(\frac{a}{t}\right)^4
\right]g f_\phi f_w
\]

The complete stress-intensity expression commonly takes the form:

\[
K_I=\sigma\sqrt{\frac{\pi a}{Q}}F
\]

The exact definitions of \(Q\), \(g\), \(f_\phi\), \(f_w\), and any loading-specific terms must be taken from the selected Newman–Raju formulation and loading case.

For the present task:

\[
\boxed{G_2=M_2=-0.54+\frac{0.89}{0.2+(a/c)}}
\]

with:

\[
0<a/c\leq1
\]

## Critical dependency result

The expression contains \(a/c\), but it does not contain \(a/t\) or \(\phi\).

Therefore:

\[
\frac{\partial M_2}{\partial(a/t)}=0
\]

and, for this coefficient expression,

\[
\frac{\partial M_2}{\partial\phi}=0
\]

Thus:
- Changing \(a/t\) does not change \(M_2\).
- Changing \(a/c\) does change \(M_2\).
- Choosing \(\phi=90^\circ\) does not change \(M_2\).
- A table indexed by both \(a/t\) and \(a/c\) will repeat the same \(M_2\) value across all \(a/t\) values at a fixed \(a/c\).

This is a mathematical consequence of the equation, not a numerical approximation.

# 7. Python implementation: table, plot, and verification

The code below:
- defines the \(M_2\) equation;
- validates the aspect-ratio domain;
- creates a two-variable table over \(a/t\) and \(a/c\);
- exports a CSV file;
- plots \(M_2\) against \(a/c\);
- checks positivity, monotonic decrease, and independence from \(a/t\).

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Geometry ratio ranges
a_over_t = np.round(np.arange(0.1, 0.81, 0.1), 2)
a_over_c = np.round(np.arange(0.1, 1.01, 0.1), 2)

def calculate_M2(a_over_c):
    # Newman-Raju M2 for 0 < a/c <= 1
    if not 0 < a_over_c <= 1:
        raise ValueError("a/c must be in (0, 1].")
    return -0.54 + 0.89 / (0.2 + a_over_c)

# Build the two-variable table.
rows = []
for x in a_over_t:
    for y in a_over_c:
        rows.append({
            "a/t": x,
            "a/c": y,
            "phi_deg": 90,
            "M2": calculate_M2(y)
        })

df = pd.DataFrame(rows)

print("NEWMAN-RAJU M2 TABLE")
display(df)

# Save the table.
csv_filename = "Newman_Raju_M2_Table.csv"
df.to_csv(csv_filename, index=False)
print(f"Saved: {csv_filename}")

# Plot M2 against a/c.
y_values = np.linspace(0.1, 1.0, 300)
M2_values = np.array([calculate_M2(y) for y in y_values])

plt.figure(figsize=(9, 6))
plt.plot(y_values, M2_values, linewidth=2.5, label="Newman-Raju M2")
plt.scatter(
    a_over_c,
    [calculate_M2(y) for y in a_over_c],
    label="Tabulated values"
)
plt.xlabel("Crack aspect ratio, a/c")
plt.ylabel("Quadratic coefficient, M2")
plt.title("Newman-Raju Quadratic Coefficient M2")
plt.xlim(0.1, 1.0)
plt.grid(True)
plt.legend()
plt.tight_layout()

plot_filename = "Newman_Raju_M2_Plot.png"
plt.savefig(plot_filename, dpi=300)
plt.show()
print(f"Saved: {plot_filename}")

# Verification checks.
assert np.all(df["M2"] > 0), "M2 should be positive over this sampled domain."

# At each a/c, M2 must be identical for all a/t values.
for y in a_over_c:
    values = df.loc[df["a/c"] == y, "M2"].to_numpy()
    assert np.allclose(values, values[0]), f"M2 unexpectedly varies with a/t at a/c={y}"

# M2 must decrease as a/c increases over this sampled domain.
assert np.all(np.diff(M2_values) < 0), "M2 should decrease with increasing a/c."

print("All verification checks passed.")
```

# 8. Distinguish \(M_2\) from its contribution to \(F\)

Although \(M_2\) itself is independent of \(a/t\), its contribution to the polynomial geometry factor is:

\[
\boxed{M_2\left(\frac{a}{t}\right)^2}
\]

This term does depend on \(a/t\).

The distinction is fundamental:

- **Coefficient:** \(M_2\), a function of \(a/c\) only.
- **Quadratic contribution:** \(M_2(a/t)^2\), a function of both \(a/c\) and \(a/t\).

If the engineering question is how the quadratic term changes with crack depth, plot \(M_2(a/t)^2\) against \(a/t\) for several fixed values of \(a/c\).

At \(\phi=90^\circ\), the usual Newman–Raju angular expression has \(f_\phi=1\); the other correction factors and their exact definitions still need to be handled according to the selected formulation. For a sufficiently wide plate, the finite-width correction may approach unity, but the small-width-effect assumption should be quantified rather than used as an unexamined omission.

## Code: plot the quadratic contribution

```python
# Plot the quadratic contribution M2 * (a/t)^2
a_over_t_fine = np.linspace(0.05, 0.8, 250)
aspect_ratios = [0.2, 0.5, 1.0]

plt.figure(figsize=(9, 6))

for ac in aspect_ratios:
    m2 = calculate_M2(ac)
    contribution = m2 * a_over_t_fine**2
    plt.plot(
        a_over_t_fine,
        contribution,
        linewidth=2,
        label=f"a/c = {ac:.1f}"
    )

plt.xlabel("Normalized crack depth, a/t")
plt.ylabel(r"Quadratic contribution, $M_2(a/t)^2$")
plt.title("Newman-Raju Quadratic-Term Contribution")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

# 9. Engineering limitations and validation discipline

The coefficient equation is only one part of an empirical stress-intensity solution.

Before using the results in a design or failure assessment:

1. Confirm the exact Newman–Raju publication, edition, and equation branch.
2. Confirm the definitions of plate width and any half-width convention.
3. Confirm whether the loading is remote tension, bending, or a combination.
4. Include the appropriate \(Q\), angular, free-surface, and finite-width factors for the complete \(K_I\) calculation.
5. Confirm that the geometry lies within the model's validated range.
6. Distinguish numerical implementation accuracy from the empirical model's accuracy relative to physical or finite-element results.

The original Newman–Raju work is based on three-dimensional finite-element analyses and provides empirical stress-intensity equations for surface cracks. A numerical implementation can reproduce an equation precisely while the empirical equation itself still has a finite validation range and modeling error.

Do not use \(M_2\) alone as a complete \(K_I\) solution.
